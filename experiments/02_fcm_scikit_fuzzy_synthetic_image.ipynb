```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

# Charger l'image
img = cv2.imread('../data/TESTsat.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Redimensionner l'image si trop grande (facultatif)
# gray = cv2.resize(gray, (200, 200))

# Aplatir l'image pour avoir un vecteur 1D
pixels = gray.flatten().astype(np.float64)

# Reshape pour FCM : 1 ligne, N pixels (scikit-fuzzy attend ce format)
data = np.expand_dims(pixels, axis=0)

# Appliquer Fuzzy C-Means clustering
n_clusters = 2  # Choisis le nombre de classes que tu veux
cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    data, c=n_clusters, m=2, error=0.005, maxiter=1000, init=None)

# Pour chaque pixel, choisir la classe avec le plus grand degré d'appartenance
cluster_labels = np.argmax(u, axis=0)

# Remodeler l'image segmentée
segmented = cluster_labels.reshape(gray.shape)

# Afficher la segmentation
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Image en niveaux de gris")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(segmented, cmap='jet')  # 'jet' donne une bonne visibilité des classes
plt.title("Segmentation FCM ({} classes)".format(n_clusters))
plt.axis("off")

plt.tight_layout()
plt.show()
```
